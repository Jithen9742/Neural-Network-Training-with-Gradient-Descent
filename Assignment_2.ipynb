```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Step 2: Sample data
x1 = 3
x2 = 2
yactual = 17

# Step 3: Initialize weights randomly between 1 and 10
w1 = np.random.uniform(1, 10)
w2 = np.random.uniform(1, 10)

# Hyperparameters
learning_rate = 0.01
epochs = 20

# Step 4: Training the model using feedforward and backpropagation
errors = []
predictions = []

for epoch in range(epochs):
    # Feedforward step: calculate predicted output (ypred)
    ypred = x1 * w1 + x2 * w2

    # Compute the error (squared error)
    error = (yactual - ypred) ** 2
    errors.append(error)

    # Save predicted values for plotting later
    predictions.append(ypred)

    # Backpropagation: Compute gradients and update weights
    d_error_d_ypred = -2 * (yactual - ypred)  # Derivative of error wrt ypred
    d_ypred_d_w1 = x1  # Derivative of ypred wrt w1
    d_ypred_d_w2 = x2  # Derivative of ypred wrt w2

    # Gradients
    grad_w1 = d_error_d_ypred * d_ypred_d_w1
    grad_w2 = d_error_d_ypred * d_ypred_d_w2

    # Update weights
    w1 = w1 - learning_rate * grad_w1
    w2 = w2 - learning_rate * grad_w2

# Step 5: Plotting the results
# Plot error over epochs
plt.subplot(1, 2, 1)
plt.plot(range(epochs), errors, label='Error')
plt.title('Error vs Epochs')
plt.xlabel('Epochs')
plt.ylabel('Error')

# Plot predicted output over epochs
plt.subplot(1, 2, 2)
plt.plot(range(epochs), predictions, label='Predicted Output (ypred)', color='orange')
plt.axhline(y=yactual, color='r', linestyle='--', label='Actual Output (yactual)')
plt.title('Predicted Output vs Epochs')
plt.xlabel('Epochs')
plt.ylabel('Predicted Output (ypred)')
plt.legend()

plt.tight_layout()
plt.show()

# Final output
print(f"Final predicted ypred: {ypred}")
print(f"Final weights: w1 = {w1}, w2 = {w2}")
```
